```python
import matplotlib.pyplot as plt
import pandas as pd
import sklearn
import sklearn.linear_model
import numpy as np
```

```python
df = pd.read_csv('./data/birthweight.csv')
```

```python
df.head()
```

```python
# Missing data 
df.isna().mean()
```

```python
df.groupby(df['Estimated fetal weight (g) - 37 weeks'].isna()).mean()
```

```python
df_37_missing = df[df['Estimated fetal weight (g) - 37 weeks'].isna()].dropna(subset=['Estimated fetal weight (g) - 32 weeks', 'Weight_neonate (g)'])
df_37_non_missing = df[~df['Estimated fetal weight (g) - 37 weeks'].isna()].dropna(subset=['Estimated fetal weight (g) - 32 weeks', 'Weight_neonate (g)'])

# For each one, do a regresssion of the birthweight from the 32 weeks EFW. Compare the slopes.
lm_missing = sklearn.linear_model.LinearRegression()
lm_non_missing = sklearn.linear_model.LinearRegression()

lm_missing.fit(df_37_missing['Estimated fetal weight (g) - 32 weeks'].values.reshape(-1, 1), df_37_missing['Weight_neonate (g)'].values)
lm_non_missing.fit(df_37_non_missing['Estimated fetal weight (g) - 32 weeks'].values.reshape(-1, 1), df_37_non_missing['Weight_neonate (g)'].values)

print("Slope for 32 weeks EFW (missing data):", lm_missing.coef_[0])
print("Slope for 32 weeks EFW (non-missing data):", lm_non_missing.coef_[0])

```
